## Waveform + spectrum

```
def show(name):
    sig = voices.render_event(SOUNDS[name]).astype(np.float64)
    fig, ax = plt.subplots(1, 2, figsize=(11, 3))
    ax[0].plot(np.arange(len(sig)) / SR, sig); ax[0].set_title(name + ' waveform')
    mag = np.abs(np.fft.rfft(sig)); freqs = np.fft.rfftfreq(len(sig), 1 / SR)
    ax[1].semilogx(freqs[1:], 20 * np.log10(mag[1:] + 1e-9)); ax[1].set_title('spectrum (dB)')
    ax[1].set_xlim(20, SR / 2)
    plt.tight_layout(); plt.show()
show('stop')
```
